```python
print("==================================================")
print("   SMARTCAMPUS UTILITY & ACCESS PASS GENERATOR   ")
print("==================================================")

# Inputs and User Type
cat = int(input("Select User Category (1: Student, 2: Faculty/Staff): "))

base_fee = 0
discount = 0
surcharge = 0

if cat == 1:
    sub = input("Enter Sub-Category (UG / PG): ")
    if sub == "UG":
        base_fee = 500
    elif sub == "PG":
        base_fee = 350

    cgpa = float(input("Enter Student CGPA (0.0 - 10.0): "))
    if cgpa < 0.0 or cgpa > 10.0:
        print("[ERROR]: CGPA must be between 0.0 and 10.0")
        exit()

    if cgpa >= 8.5:
        discount = 0.20 * base_fee
    elif cgpa >= 7.5:
        discount = 0.10 * base_fee

elif cat == 2:
    sub = input("Enter Sub-Category (Resident / Visiting): ")
    if sub == "Resident":
        base_fee = 800
    elif sub == "Visiting":
        base_fee = 1200

    exp = int(input("Enter Years of Service: "))
    if exp < 0:
        print("[ERROR]: Years of Service must be non-negative integer.")
        exit()

    if exp > 10:
        discount = 0.15 * base_fee

# Parking
park = int(input("Select Parking Permit (0: None, 2: Two-Wheeler, 4: Four-Wheeler): "))
park_fee = 0

if park == 2:
    park_fee = 200
elif park == 4:
    park_fee = 600
    if cat == 1:
        surcharge = 150

# Electricity
units = float(input("Enter Monthly Electricity Consumption (in kWh): "))
if units < 0:
    print("[ERROR]: Electricity Units consumed cannot be negative.")
    exit()

if units <= 100:
    fixed = 50
    e_bill = (units * 3.0) + fixed
elif units <= 300:
    fixed = 100
    e_bill = (100 * 3.0) + ((units - 100) * 5.0) + fixed
elif units <= 500:
    fixed = 150
    e_bill = (100 * 3.0) + (200 * 5.0) + ((units - 300) * 7.5) + fixed
else:
    fixed = 250
    e_bill = (100 * 3.0) + (200 * 5.0) + (200 * 7.5) + ((units - 500) * 10.0) + fixed

# Totals
pass_total = (base_fee - discount) + park_fee + surcharge
total = pass_total + e_bill

# Output
print("\n--------------------------------------------------")
print("          CALCULATED INVOICE BREAKDOWN           ")
print("--------------------------------------------------")
print("Base Access Pass Fee      : ₹", float(base_fee))
if discount > 0:
    print("Merit Discount            :- ₹", float(discount))
if park_fee > 0:
    print(f"Parking Fee ({park}-Wheeler)    : ₹", float(park_fee))
if surcharge > 0:
    print("Student Peak Surcharge    : ₹", float(surcharge))
print("Net Pass & Parking Total  : ₹", float(pass_total))
print("--------------------------------------------------")
print(f"Electricity Bill ({int(units)} kWh) : ₹ {e_bill:.2f} (Slab calculated + Fixed Charge )")
print("--------------------------------------------------")
print("TOTAL MONTHLY PAYABLE     : ₹", float(total))
```
